# Question 1: What is hypothesis testing in statistics?
-  Hypothesis testing is a formal statistical procedure used to make inferences about a population parameter based on sample data. It helps determine whether there is sufficient statistical evidence to support or reject a proposed assumption about the population.

-  In hypothesis testing, a researcher begins by formulating two competing hypotheses:

   - The Null Hypothesis (H₀), which represents the default assumption or status quo (e.g., no effect, no difference).

   - The Alternative Hypothesis (H₁ or Hₐ), which represents the claim being tested or the presence of an effect or difference.

   - Using sample data, a test statistic is calculated and compared against a probability distribution to determine the p-value, which quantifies the likelihood of observing the data assuming the null hypothesis is true. This p-value is then compared to a predefined significance level (α), commonly set at 0.05.

   - If the p-value is less than or equal to α, the null hypothesis is rejected, indicating that the observed result is statistically significant. Otherwise, the null hypothesis is not rejected.

- In data science, hypothesis testing is widely used for:

  - Model validation and comparison

  - A/B testing and experimentation

  - Feature selection

  - Business decision-making under uncertainty

- Overall, hypothesis testing provides a rigorous, data-driven framework for drawing conclusions and minimizing decision-making errors.

# Question 2: What is the null hypothesis, and how does it differ from the alternative hypothesis?
- The null hypothesis (H₀) is a statistical statement that assumes there is no effect, no difference, or no relationship between variables in a population. It represents the default or baseline assumption and is formulated so that it can be tested and potentially rejected using sample data.

- The alternative hypothesis (H₁ or Hₐ) is the statement that contradicts the null hypothesis. It represents the presence of an effect, a difference, or a relationship and reflects what the researcher aims to provide evidence for through hypothesis testing.

- Example:- In an A/B testing scenario:

  -  H₀: There is no difference in conversion rates between Version A and Version B.

  -  H₁: There is a difference in conversion rates between Version A and Version B.
  

# Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.
- The significance level, denoted by α (alpha), is a predefined threshold used in hypothesis testing to determine whether the results of a statistical test are statistically significant. It represents the maximum probability of committing a Type I error, which occurs when the null hypothesis is incorrectly rejected even though it is true.

-  Commonly used significance levels are 0.05, 0.01, and 0.10, depending on the acceptable level of risk in decision-making. A lower value of α indicates a more stringent test, reducing the chance of false positives but increasing the risk of Type II errors.

-  Role in Decision-Making:
   - After calculating the p-value from sample data:

   - If p-value ≤ α, the null hypothesis is rejected, indicating strong evidence against it.

   - If p-value > α, the null hypothesis is not rejected, suggesting insufficient evidence to support the alternative hypothesis.

# Question 4: What are Type I and Type II errors? Give examples of each.
- - In hypothesis testing, we make decisions based on sample data.
However, because we work with samples and not the full population, errors can happen. These errors are classified as:

   -  Type I Error (False Positive) occurs when we reject the null hypothesis (H₀), even though it is actually true.

Example:

  -  A COVID test says the person is positive, but actually they are not.

Symbol: α (alpha)

- Impact on Decision Making:

   - We make a decision thinking there is an effect/difference when actually there is none.

   - It can lead to unnecessary actions or wrong conclusions.

Higher Type I error = low reliability of results.

-  Type II Error (False Negative) occurs when we fail to reject the null hypothesis (H₀), even though it is actually false.

Example:

   -  A COVID test says the person is negative, but actually they are infected.

Symbol: β (beta)

- Impact on Decision Making:

   - We miss detecting a real effect/difference.

   - It can lead to harmful delays, missed risks, or missed opportunities.

   - High Type II error means the test or experiment is not powerful enough.

# Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.
- -  A Z-test is used when the sample size is large (n > 30) or the population standard deviation is known.
A T-test is used when the sample size is small (n ≤ 30) and the population standard deviation is unknown.
Z-test uses the Z-distribution, while T-test uses the T-distribution, which is wider to handle small samples.
As sample size increases, the T-distribution becomes similar to the Z-distribution.

- Example:

   - If you have a small sample of 20 students and you don’t know the population standard deviation, you use a T-test.
   - If you have a large sample of 200 people and know the population standard deviation, you use a Z-test.

# Question 8: Write a Python script to simulate data from a normal distribution and calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Step 1: Simulate data from a normal distribution
np.random.seed(42)  # For reproducibility
mu = 50             # Population mean
sigma = 5           # Population standard deviation
n = 100             # Sample size

# Generate sample data
data = np.random.normal(loc=mu, scale=sigma, size=n)

# Step 2: Calculate sample mean and standard error
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
standard_error = sample_std / np.sqrt(n)

# Step 3: Calculate 95% confidence interval
confidence_level = 0.95
z_score = stats.norm.ppf((1 + confidence_level) / 2)  # two-tailed
margin_of_error = z_score * standard_error

ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

# Print results
print(f"Sample Mean: {sample_mean:.2f}")
print(f"95% Confidence Interval: ({ci_lower:.2f}, {ci_upper:.2f})")

# Step 4: Plot the data
plt.figure(figsize=(10,5))
plt.hist(data, bins=15, color='skyblue', edgecolor='black', alpha=0.7)
plt.axvline(sample_mean, color='red', linestyle='--', label=f'Mean = {sample_mean:.2f}')
plt.axvline(ci_lower, color='green', linestyle='--', label=f'95% CI Lower = {ci_lower:.2f}')
plt.axvline(ci_upper, color='green', linestyle='--', label=f'95% CI Upper = {ci_upper:.2f}')
plt.title('Histogram of Simulated Normal Data')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

# Question 9: Write a Python function to calculate the Z-scores from a dataset and visualize the standardized data using a histogram. Explain what the Z-scores represent in terms of standard deviations from the mean.

```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Define a function to calculate Z-scores
def calculate_z_scores(data):
    """
    This function takes a dataset as input and returns the Z-scores.
    Z-score = (value - mean) / standard deviation
    """
    mean = np.mean(data)
    std = np.std(data, ddof=1)  # sample standard deviation
    z_scores = (data - mean) / std
    return z_scores

# Step 2: Sample dataset
data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
        50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
        50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
        50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Step 3: Calculate Z-scores
z_scores = calculate_z_scores(data)

# Step 4: Visualize Z-scores with a histogram
plt.figure(figsize=(10,5))
plt.hist(z_scores, bins=10, color='lightcoral', edgecolor='black', alpha=0.7)
plt.axvline(0, color='blue', linestyle='--', label='Mean (Z=0)')
plt.title('Histogram of Z-scores (Standardized Data)')
plt.xlabel('Z-score')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```
